1.What is the difference between interpreted and compiled languages?

Ans. Interpreted and compiled languages are two distinct approaches to translating and executing programming code. Here's a breakdown of their key differences


Compiled Languages:

- Translation: The entire source code is translated into machine code (or an intermediate bytecode) before the program is executed. This translation is done by a program called a compiler.
-Execution: The resulting machine code (or bytecode) is then executed directly by the computer's central processing unit (CPU) or a virtual machine.
-Speed: Compiled programs generally run faster because the translation is done only once, and the resulting code is often optimized for the target hardware.
-Portability: The compiled code is often platform-specific. To run the program on a different operating system or hardware architecture, it usually needs to be recompiled for that specific platform.
-Error Detection: Compilers typically perform static analysis and can catch syntax errors and some semantic errors before the program runs.
-Development Speed: The compilation step can make the development process slightly slower as you need to compile the code every time you make changes before testing.
-Examples: C, C++, C#, Go, and Rust are common examples of compiled languages. Java is a bit of a hybrid; it's compiled to bytecode, which is then executed by the Java Virtual Machine (JVM).

Interpreted Languages:

- Translation: The source code is translated and executed line by line at runtime by a program called an interpreter. There's no separate compilation phase.
-Execution: The interpreter reads a line of code, translates it into an intermediate form or directly into machine instructions, and then executes it. This process is repeated for each line of code.
-Speed: Interpreted programs generally run slower than compiled programs because each line of code needs to be translated every time it's executed.
-Portability: Interpreted languages are often more portable. As long as there's an interpreter available for a specific operating system, the same source code can usually run on it without modification.
-Error Detection: Errors in interpreted languages are typically detected at runtime, only when the line of code containing the error is executed.
-Development Speed: Interpreted languages often offer a faster development cycle because you can make changes and run the code immediately without a compilation step, which is beneficial for rapid prototyping and debugging.
-Examples: Python, JavaScript, Ruby, and PHP are popular examples of interpreted languages

6. What is the difference between import and from ... import in Python?

Ans.  the import and from ... import statements in Python are both used to bring modules and their contents into your current script's namespace, but they differ in how they do so and how you subsequently access those contents. Let's break down the distinctions:

1. import <module_name>

What it does: This statement imports the entire specified module. The module becomes an object in your current namespace, and you access its contents (functions, classes, variables, etc.) using the module name as a prefix, followed by a dot (.).

Namespace Impact: It imports the module itself into the current namespace. The names within the module (unless explicitly imported otherwise) do not directly pollute the current namespace.

Accessing Members: You access the members of the module using the dot notation: <module_name>.<member_name>.

Readability (Potential Trade-off): While it requires more typing to access members, it can make the code more readable by explicitly indicating which module a particular function or class comes from, especially when you are using multiple modules with potentially overlapping names.

Example:

```python
import math

print(math.sqrt(16))  # Accessing the sqrt function using math.sqrt
print(math.pi)       # Accessing the pi constant using math.pi
```

2. from <module_name> import <name(s)>

What it does: This statement imports specific names (functions, classes, variables) directly from the specified module into your current namespace. You can import one or more names, separated by commas, or use a wildcard (*) to import all public names (though using * is generally discouraged).

Namespace Impact: The imported names become directly available in your current namespace, without needing the module name as a prefix.

Accessing Members: You access the imported names directly, without the module name prefix.

Readability (Potential Trade-Off): It can make the code more concise as you don't need to type the module name repeatedly. However, if you import many names or if there are name collisions with existing names in your script or from other modules, it can make the code harder to read and understand where a particular name originates.

Examples:

```python
from math import sqrt
print(sqrt(25))  # Accessing sqrt directly
```

7.  How can you handle multiple exceptions in Python?

Ans.  handling multiple exceptions in Python is a common and important practice to make your code more robust. You can address various potential errors in different ways within a single try block. Here are several ways to handle multiple exceptions:

1. Separate except Blocks:

This is the most straightforward and often the most readable way to handle different types of exceptions with specific code for each. You simply have multiple except clauses following a single try block, each catching a different exception type.

```python
try:
    numerator = int(input("Enter the numerator: "))
    denominator = int(input("Enter the denominator: "))
    result = numerator / denominator
    print(f"The result is: {result}")
except ValueError:
    print("Invalid input. Please enter numbers only.")
except ZeroDivisionError:
    print("Error: Cannot divide by zero.")
except FileNotFoundError:
    print("Error: The specified file was not found.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

2. Catching Multiple Exceptions in a Single except Block (using a tuple):

If you want to handle several different exception types with the same block of code, you can specify them as a tuple within a single except clause.

```python
try:
    value = int(input("Enter a number between 1 and 10: "))
    if value < 1 or value > 10:
        raise ValueError("Value out of range")
    result = 10 / (value - 5)
    print(f"Result: {result}")
except (ValueError, ZeroDivisionError) as e:
    print(f"Error encountered: {e}. Please check your input.")
except TypeError:
    print("A type error occurred.")
```

3. Using a Base Exception Class:

You can catch a broader category of exceptions by specifying a base class in the except clause. For example, ArithmeticError is a base class for exceptions related to arithmetic operations like ZeroDivisionError and OverflowError.

```python
try:
    numerator = int(input("Enter the numerator: "))
    denominator = int(input("Enter the denominator: "))
    result = numerator / denominator
    print(f"The result is: {result}")
except ArithmeticError as e:
    print(f"An arithmetic error occurred: {e}")
except ValueError:
    print("Invalid input. Please enter numbers only.")
```

18. What is the difference between os.fork() and multiprocessing in Python?

Ans. os.fork() and the multiprocessing module in Python both provide ways to achieve parallelism by creating new processes, but they operate on fundamentally different principles and have distinct characteristics, especially concerning how they handle resources and their portability.

Here's a breakdown of the key differences:

1. Underlying Mechanism:

os.fork(): This is a low-level system call that directly creates a new process by duplicating the existing (parent) process. The child process gets an exact copy of the parent's memory space at the moment of the fork(). This includes code, data, open files, and even the program counter. After the fork, both the parent and child processes continue executing from the instruction following the fork() call, but they have distinct process IDs (PIDs).
multiprocessing: This is a higher-level module that provides a more Pythonic and platform-independent way to create and manage processes. It typically uses mechanisms like fork() (on Unix-like systems), spawn (on Windows and macOS by default), or forkserver (available on some Unix systems) behind the scenes to create new processes. The multiprocessing module generally starts a fresh Python interpreter in the new process, which then executes the target function.
2. Memory Sharing:

os.fork(): Because the child process gets a copy-on-write duplicate of the parent's memory, initially, both processes share the same memory pages. However, if either process modifies a shared memory page, a new copy of that page is created for the modifying process. This "copy-on-write" mechanism can be efficient but can also lead to complexities in sharing data between processes. Explicit mechanisms like shared memory segments or pipes are usually needed for inter-process communication (IPC).
multiprocessing: By default, the multiprocessing module avoids direct sharing of the parent's memory in the same way as fork(). When using the spawn or forkserver start methods, each new process gets its own independent memory space. If you need to share data, multiprocessing provides specific tools like Queue, Pipe, Value, Array, and Manager objects, which handle the underlying IPC mechanisms in a more managed way.

3. Resource Sharing:

os.fork(): The child process inherits most of the parent's resources, such as open file descriptors. This can be both convenient and potentially problematic if not managed carefully (e.g., both processes might try to write to the same file concurrently without proper synchronization).
multiprocessing: The multiprocessing module offers more control over resource sharing. When using spawn or forkserver, the child process starts with a clean slate and only explicitly passed resources are available. This can lead to more robust and predictable behavior.

4. Portability:

os.fork(): os.fork() is primarily available on Unix-like systems (Linux, macOS, etc.). It is not available on Windows. Code that relies directly on os.fork() will not be portable to Windows.
multiprocessing: The multiprocessing module is designed to be cross-platform. It abstracts away the underlying process creation mechanism, so your multiprocessing code generally works on Windows, macOS, and Linux without significant changes. It automatically chooses the appropriate start method (spawn on Windows and macOS by default, fork on most Unix-like systems if available and safe).

5. Complexity and Ease of Use:

os.fork(): Working directly with os.fork() can be more complex, especially when it comes to managing shared resources and ensuring proper synchronization between parent and child processes. You often need to handle signals and manage the lifecycle of the child process explicitly.
multiprocessing: The multiprocessing module provides a higher-level API with classes like Process, Pool, Queue, Pipe, and synchronization primitives (locks, semaphores, etc.) that make it easier to create and manage concurrent processes and handle inter-process communication in a more structured way.

19. What is the importance of closing a file in Python?

Ans. Ah, closing a file in Python is more important than it might initially seem! While Python's garbage collection will eventually close files when they are no longer referenced, explicitly closing them using the close() method or, even better, a with statement, is crucial for several reasons:

Resource Management: Operating systems have limits on the number of files that a process can have open simultaneously. If you open many files without closing them, you can quickly exceed this limit, leading to errors and potentially crashing your program or even other parts of the system. Closing files releases these system resources, making them available for other operations.

Data Integrity: When you write data to a file, it's not always immediately written to the physical disk. Instead, the operating system often buffers the data in memory for efficiency. Calling close() forces the operating system to flush these buffers, ensuring that all the data you intended to write is actually saved to the file. If your program exits or crashes before you close the file, some of your data might be lost or corrupted.

Preventing Data Corruption: If a file is left open and another process (or even another part of the same program) tries to access or modify it, it can lead to data corruption or unexpected behavior. Closing the file ensures that it's in a consistent state and available for other operations.

File Locking: Operating systems often implement file locking mechanisms. When a file is open for writing, it might be locked to prevent other processes from modifying it simultaneously. Failing to close the file can leave it locked, preventing other programs or even the same program from accessing it later.

Portability and Interoperability: While Python's garbage collection is generally reliable, the exact timing of when a file object is deallocated and closed can be unpredictable and might vary across different Python implementations or operating systems. Explicitly closing files ensures consistent behavior across different environments.

Use Cases:
Reading a file line by line, which is memory-efficient for large files as you process one line at a time.
Parsing files where data is structured on a per-line basis (e.g., CSV files, log files).

```python
#2.  Write a Python program to read the contents of a file and print each line

def read_and_print_file(filename):
    """
    Reads the contents of a file and prints each line.

    Args:
        filename (str): The name of the file to read.
    """
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            for line in file:
                # Print each line, removing any leading/trailing whitespace
                print(line.strip())
    except FileNotFoundError:
        print(f"Error: File not found: {filename}")
    except Exception as e:
        print(f"An error occurred while reading the file: {e}")

if __name__ == "__main__":
    # Get the filename from the user.  Important to get input
    # to increase reusability of the code.
    filename = input("Enter the name of the file to read: ")
    read_and_print_file(filename)

```

```python
#4.  Write a Python script that reads from one file and writes its content to another file

def copy_file_content(source_filename, destination_filename):
    """
    Reads the content of one file and writes it to another file.

    Args:
        source_filename (str): The name of the file to read from.
        destination_filename (str): The name of the file to write to.
    """
    try:
        # Open the source file for reading with utf-8 encoding
        with open(source_filename, 'r', encoding='utf-8') as source_file:
            # Read the entire content of the source file
            file_content = source_file.read()

        # Open the destination file for writing with utf-8 encoding
        # If the file exists, it will be overwritten.
        with open(destination_filename, 'w', encoding='utf-8') as destination_file:
            # Write the content to the destination file
            destination_file.write(file_content)

        print(f"Successfully copied content from '{source_filename}' to '{destination_filename}'")

    except FileNotFoundError:
        print(f"Error: Source file not found: '{source_filename}'")
    except Exception as e:
        print(f"An error occurred: {e}")

if __name__ == "__main__":
    # Get the source and destination filenames from the user
    source_filename = input("Enter the name of the source file: ")
    destination_filename = input("Enter the name of the destination file: ")

    # Call the function to copy the file content
    copy_file_content(source_filename, destination_filename)
```

```python
#9.  How can you read a file line by line and store its content in a list in Python?

def read_file_to_list(filename):
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            lines = file.readlines()  # Read all lines into a list
            return lines
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
        return []  # Return an empty list in case of error
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
        return []

if __name__ == "__main__":
    filename = input("Enter the filename: ")
    file_content = read_file_to_list(filename)

    # Now you can work with the file_content list
    # Example: Print each line
    for line in file_content:
        print(line.strip())

```

```python
#17.  Write a Python program to create and write a list of numbers to a file, one number per line:

def write_numbers_to_file(filename, numbers):
    try:
        with open(filename, 'w', encoding='utf-8') as file:
            for number in numbers:
                file.write(str(number) + '\n')  # Write each number followed by a newline
            print(f"Numbers written to '{filename}' successfully.")
    except Exception as e:
        print(f"An error occurred: {e}")

if __name__ == "__main__":
    filename = input("Enter the filename: ")
    numbers = [1, 2, 3, 4, 5]  # Example list of numbers
    write_numbers_to_file(filename, numbers)
```

```python
#18.  How would you implement a basic logging setup that logs to a file with rotation after 1MB?

import logging
from logging.handlers import RotatingFileHandler

def setup_logging(log_file, max_bytes=1048576, backup_count=5):
    """Sets up basic logging to a file with rotation.

    Args:
        log_file: The path to the log file.
        max_bytes: Maximum size of the log file in bytes before rotation
                   (default: 1 MB).
        backup_count: Number of backup log files to keep (default: 5).
    """

    logger = logging.getLogger(__name__)
    logger.setLevel(logging.DEBUG)  # Set desired logging level

    # Create a rotating file handler
    handler = RotatingFileHandler(
        log_file,
        maxBytes=max_bytes,
        backupCount=backup_count
    )

    formatter = logging.Formatter(
        '%(asctime)s - %(levelname)s - %(message)s'
    )
    handler.setFormatter(formatter)

    logger.addHandler(handler)
    return logger

if __name__ == "__main__":
    logger = setup_logging('my_log.log')

    # Example logging messages
    logger.debug('This is a debug message.')
    logger.info('This is an info message.')
    logger.warning('This is a warning message.')
    logger.error('This is an error message.')
    logger.critical('This is a critical message.')
```

```python
#21. Write a Python program that reads a file and prints the number of occurrences of a specific word:

import re

def count_word_occurrences(filename, word):
    """
    Counts the occurrences of a specific word in a file.

    Args:
        filename: The path to the file.
        word: The word to count occurrences of.

    Returns:
        The number of occurrences of the word in the file.
    """
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            content = file.read()
            # Use regex to find all whole word matches (case-insensitive)
            occurrences = len(re.findall(r'\b' + word + r'\b', content, re.IGNORECASE))
            return occurrences
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
        return 0
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
        return 0

if __name__ == "__main__":
    filename = input("Enter the filename: ")
    word = input("Enter the word to count: ")

    count = count_word_occurrences(filename, word)
    print(f"The word '{word}' appears {count} times in the file.")
```

```python
#22. How can you check if a file is empty before attempting to read its contents?

import os

def is_file_empty(filename):
    """Checks if a file is empty.

    Args:
        filename: The path to the file.

    Returns:
        True if the file is empty, False otherwise.
    """
    try:
        return os.stat(filename).st_size == 0
    except OSError:
        # File might not exist or is inaccessible
        return False

if __name__ == "__main__":
    filename = input("Enter the filename: ")
    if is_file_empty(filename):
        print(f"The file '{filename}' is empty.")
    else:
        # Proceed with reading the file contents
        print(f"The file '{filename}' is not empty.")
```
